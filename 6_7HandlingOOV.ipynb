- We will now look at the technique for **handling out-of-vocabulary words**. Let's look a little bit about **limitations of Word2vec**.

We have two issues:
1. The first issue is **homonyms**. These are words **that have identical spelling,** **but different meanings**. For example, in the sentence, "***Can*** I have a ***can*** of soda**?" We have **two instances of the word can, but they mean different things**.

 However, with Word2vec, they would map to the **same encoding**. Without any **context** it's unclear which can refers to what.

2. Another **issue with Word2vec** is how **to handle out-of-vocabulary words**. We simply **don't have an encoding for unseen words**, so what we do is that we **have to map them to just some kind of common encoding for an unknown word.**

However, it's often **the case that even unknown words look similar to some known words**. Therefore, we can make use of that and do something more intelligent than just map them to unknown. We will look at the solution for this second problem, the unknown word problem, using something **known as Wordpieces**.

**Using the Wordpieces scheme**, we will **encode either full words or word fragments**, so our vocabulary will consist of both fragments as well as full words. The smallest fragment will be as small as individual characters.

**A specific word will map to one or more Wordpieces**. In the extreme case, each piece is a character spelling the word. Using these schemes allows a unique encoding, even if the word hasn't been seen during training.

So what we will **do is to break words into pieces** and then **we learn embeddings**. We can **use any embedding method on these pieces**. For example, **Word2vec**, **Glove**, **or simply train it jointly with a language model**. Let's look at an example from the original paper introduces Wordpieces.

If we have the sentence, "**Jet makers feud over seat width with big orders at stake."**

And we assume that **jet and feud are unseen during training**. we would then break **these words in two pieces**. So **Jet might become J** **and et, and feud might become fe and ud**.
